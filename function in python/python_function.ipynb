```python
# default mutable argument
def add_item_good(item, items =None):
  if items is None:
    items = []
  items.append(item)
  return items
print(add_item_good("apple"))
print(add_item_good("banana"))
print(add_item_good("guava"))
```

```python
# *arguments_variable number of positional arguments

def add_all(*number):
  print(f"received: {number}, (type: {type(number).__name__})")
  return sum(number)

print(f"sum ={add_all(1,2,3,4,5,6)}")
```

```python
# **kwargs_variable number of keyword arguments

def print_info(**details):
  print(f"received : {details}, (type:{type(details).__name__})")
  for key,value in details.items():
    print(f"{key}:{value}")

print_info(name="parvej",age="29",city="madina")
```

```python
# nested function

def process_grades(grades):
  def get_letter(score):
    if score >90: return "A"
    elif score >80: return "B"
    elif score >70: return "C"
    elif score >60: return "D"
    else: return "F"
  for score in grades:
    print(f"{score}:{get_letter(score)}")
```

```python
process_grades([56,89,90,98,97,78])
```

```python
# lambda with sorting

students =[
    {"name":"parvej","grade": 99},
    {"name":"kamal","grade": 80},
    {"name":"riyad","grade": 70},
    {"name":"kajol","grade": 60},
]
# short by grade ascerding
sorted(students, key=lambda s: s["grade"])
sorted(students, key=lambda s: s["name"])
```

```python
# HIGHER ORDER FUNCTION
# map- apply a function to every element
numbers = [2,3,4,5,6]
list(map(lambda x:x**2,numbers))
```

```python
#filter-keep element that pass a test
numbers = [2,3,4,5,6,7,8,9]
list(filter(lambda x: x%2 == 0,numbers))
list(filter(lambda x: x%2 !=0,numbers))
```

```python
# passing function as argument
def apply_operation(func,data):
  return[func(x)for x in data]

num = [1,2,3,4,5]
def square(x):
  return x**2
```

```python
apply_operation(square,num)
```

```python
apply_operation(lambda x: x**2,num)
```
