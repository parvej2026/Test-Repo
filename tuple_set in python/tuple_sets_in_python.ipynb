```python
# union/update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.union(s2)
```

```python
# intersaction/intersaction update

s1 = {1,2,3,4}
s2 = {1,2,7}

s1.intersection(s2)
```

```python
s1.intersection_update(s2)
print(s1)
```
